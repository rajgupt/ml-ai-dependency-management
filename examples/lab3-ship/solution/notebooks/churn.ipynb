# Churn model — exploration

This notebook imports `churnkit`, the package in `src/`. It defines no
model logic of its own: anything worth keeping lives in the package, gets
tested, and runs the same way here, in CI and in production.

Run it with the project kernel:

```bash
uv run --group notebook ipython kernel install --user --name=churnkit
```

````python
from churnkit import build_features, make_customers, train
````

## The data

Same function the training job calls. Not a copy of it.

````python
df = make_customers(8_000)
df.head()
````

## The features

One definition of `spend_to_date`, in `churnkit.features`, covered by a test.
When you change it, the notebook and the training job change together.

````python
features = build_features(df)
features.columns.tolist()
````

## Train

Identical to what `churn-train` runs from the command line.

````python
result = train(df)
print(f"AUC {result.test_auc:.4f} on {result.n_test} held-out rows")
````

## What belongs where

| Belongs in the notebook | Belongs in `src/churnkit/` |
|---|---|
| Plots, one-off slices, "what does this column look like" | Anything you'd be upset to lose |
| Arguments for a decision | Anything a second person needs to reproduce |
| Scratch | Anything you want a test on |

The rule of thumb: **if you'd copy-paste it into a second notebook, it belongs
in the package.**
